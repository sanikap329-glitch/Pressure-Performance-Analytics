```python

import pandas as pd
from pathlib import Path

# Load final model ranking files
batter_file = Path("final_player_pressure_ranking.csv")
bowler_file = Path("all_bowler_pressure_rankings.csv")

if not batter_file.exists() or not bowler_file.exists():
    raise FileNotFoundError(
        "Keep both final ranking CSV files in the notebook folder."
    )

batters = pd.read_csv(batter_file)
bowlers = pd.read_csv(bowler_file)

# ---------------------------------------
# BATTER TOP 10
# ---------------------------------------

print("========== MODEL TOP 10 BATTERS ==========")

batter_score = "reliability_adjusted_score"

batter_name = next(
    c for c in ["striker_name", "batter_name", "player_name"]
    if c in batters.columns
)

batter_rank = (
    "reliability_rank"
    if "reliability_rank" in batters.columns
    else None
)

if batter_rank:
    top10_batters = batters.sort_values(batter_rank).head(10)
else:
    top10_batters = batters.sort_values(
        batter_score, ascending=False
    ).head(10)

batter_columns = [
    c for c in [
        batter_rank, batter_name, batter_score,
        "high_pressure_balls", "sample_reliability"
    ] if c and c in batters.columns
]

print(top10_batters[batter_columns].to_string(index=False))


# ---------------------------------------
# BOWLER TOP 10
# ---------------------------------------

print("\n========== MODEL TOP 10 BOWLERS ==========")

bowler_score = "reliability_adjusted_score"

bowler_name = next(
    c for c in ["bowler_name", "bowler", "player_name"]
    if c in bowlers.columns
)

bowler_rank = (
    "final_rank"
    if "final_rank" in bowlers.columns
    else None
)

if bowler_rank:
    top10_bowlers = bowlers.sort_values(bowler_rank).head(10)
else:
    top10_bowlers = bowlers.sort_values(
        bowler_score, ascending=False
    ).head(10)

bowler_columns = [
    c for c in [
        bowler_rank, bowler_name, bowler_score,
        "high_pressure_legal_balls", "sample_reliability"
    ] if c and c in bowlers.columns
]

print(top10_bowlers[bowler_columns].to_string(index=False))


# ---------------------------------------
# BASIC VALIDATION
# ---------------------------------------

print("\n========== VALIDATION ==========")

print("Total eligible batters:", len(batters))
print("Total eligible bowlers:", len(bowlers))

print("Duplicate batter names:",
      batters[batter_name].duplicated().sum())

print("Duplicate bowler names:",
      bowlers[bowler_name].duplicated().sum())

print("Batter ranking score available:",
      batter_score in batters.columns)

print("Bowler ranking score available:",
      bowler_score in bowlers.columns)

# Save verified Top 10 lists
top10_batters.to_csv("verified_model_top10_batters.csv", index=False)
top10_bowlers.to_csv("verified_model_top10_bowlers.csv", index=False)

print("\nVerified Top 10 files exported successfully.")
```
